## **Origins of Game Theory**

Game theory's origins are usually attributed to **John von Neumann**, who proved the **minimax theorem** for two-player zero-sum games in 1928. However, prior to von Neumann, several others explored games of chance and probability as early as the 16th century. The field became recognized when John von Neumann and Oskar Morgenstern published **Theory of Games and Economic Behavior** (1944), which was foundational in the establishment of game theory.

## **The Basic Components of a Game**
* **Players** - agents involved in a game
* **Actions** - the set of actions each agent can take in a game
* **Payoff** - the value each player receives for every combination of strategies
* **Information** - information each player has at the time of making their decisions



## **Why Games *and* Networks?**

An assumption in evolutionary game theory is that of a **well-mixed population**, in which every agent has an equal chance of interacting with every other agent, unhindered by geographic or social barriers. Empirically observed social networks contradict this assumption: we tend to interact with people we already know, such as our coworkers, friends, and family. Because these interactions recur, they shape our behavior differently than a one-time encounter with a stranger would.

Integrating game theory with network science offers a more realistic picture of how strategic interactions unfold. Game theory models strategic behavior through players, action sets, and payoffs, but its assumption of random encounters fails to capture real social dynamics. Network science, by contrast, provides a topological understanding of complex systems, showing how nodes connect to one another, but it does not tell us what nodes do with those connections. Combining these approaches allows us to explore how network topology and strategic behavior interact, and how that interplay shapes the outcomes we observe in social systems.

## **The Prisoner's Dilemma**

To explore these dynamics, we will focus on implementing The Prisoner's Dilemma on a series of network types that we have looked at throughout this book. In the Prisoner’s Dilemma you and your partner-in-crime are arrested and contained within separate cells. The authorities offer you and your partner the same deal; however, you are not able to communicate with them. As a player, your action set is to stay silent – cooperate with your partner — or snitch on your partner – defecting.

Here's the deal you and your partner are offered:
* If you both cooperate and choose to stay silent you'll each get a punishment of 1 year. This is the reward, $R$ for mutual cooperation.
* If you both defect and snitch on one another you'll each get a punishment of 4 years. This is the punishment, $P$ for mutual defection.
* If you defect and your partner stays silent, you'll receive 0 years while your partner receives 10. This represents the temptation to defect, $T$.
* If you stay silent and your partner defects, you'll receive 10 years while your partner receives 0 culminating in the Sucker's Payoff, $S$.

With these options, we can represent the **payoff matrix** for each player:

$$
\begin{array}{r|cc}
 & \text{P2 Cooperates} & \text{P2 Defects} \\ \hline
\text{P1 Cooperates} & (-1, -1) & (-10, 0) \\
\text{P1 Defects} & (0, -10) & (-4, -4)
\end{array}
$$

This ordering captures the dilemma: for a rational, self-interested player, defecting yields a higher payoff regardless of the opponent’s choice, your **dominant strategy**. Given that our partner has the same options, defecting is also their **dominant strategy** leading to a situation of mutual defection, $(P,P)$, is the usual outcome despite the better option of both players staying silent.

However, when we look at systems from ant colonies to human societies, cooperation flourishes (Nowak, 2006). This discrepancy highlights a fundamental issue with the simple model’s assumption: real populations are not well-mixed, as we mentioned earlier. When players interact only with their network neighbors, cooperation can survive, as Nowak and May (1992) showed.
